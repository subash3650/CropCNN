```python
import os
import shutil
import random

random.seed(42)

base = "/content/drive/MyDrive/Crop_Classification/datasets"
source_base = os.path.join(base, "agriculture-crop-images", "kag2")
final_path = os.path.join(base, "final_dataset")

sources = {
    "Paddy": "rice",
    "Wheat": "wheat",
    "Sugarcane": "sugarcane",
    "Maize": "maize"
}

for class_name, source_folder in sources.items():

    source_path = os.path.join(source_base, source_folder)
    destination = os.path.join(final_path, class_name)

    images = [
        f for f in os.listdir(source_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(images)
    selected = images[:200]

    for image in selected:
        shutil.copy2(
            os.path.join(source_path, image),
            os.path.join(destination, image)
        )

    print(f"{class_name}: {len(selected)} images copied")

print("\nFinal dataset creation completed.")
```

```python
import os

final_path = "/content/drive/MyDrive/Crop_Classification/datasets/final_dataset"

target_count = 160

for class_name in os.listdir(final_path):
    class_path = os.path.join(final_path, class_name)

    if not os.path.isdir(class_path):
        continue

    images = [
        f for f in os.listdir(class_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    # Keep exactly 160 images
    if len(images) > target_count:
        for image in images[target_count:]:
            os.remove(os.path.join(class_path, image))

    print(f"{class_name}: {min(len(images), target_count)} images kept")

print("\nBalanced dataset ready: 800 images")
```

```python
import os
import shutil
import random

random.seed(42)

base = "/content/drive/MyDrive/Crop_Classification/datasets"
source_path = os.path.join(base, "final_dataset")
split_path = os.path.join(base, "final_dataset_split")

classes = ["Maize", "Paddy", "Sugarcane", "Sunflower", "Wheat"]

# Create split folders
for split in ["train", "validation", "test"]:
    for class_name in classes:
        os.makedirs(
            os.path.join(split_path, split, class_name),
            exist_ok=True
        )

# Split each class
for class_name in classes:

    class_path = os.path.join(source_path, class_name)

    images = [
        f for f in os.listdir(class_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(images)

    train_images = images[:112]
    validation_images = images[112:136]
    test_images = images[136:160]

    splits = {
        "train": train_images,
        "validation": validation_images,
        "test": test_images
    }

    for split, split_images in splits.items():

        destination = os.path.join(
            split_path,
            split,
            class_name
        )

        for image in split_images:
            shutil.copy2(
                os.path.join(class_path, image),
                os.path.join(destination, image)
            )

    print(
        f"{class_name}: "
        f"Train={len(train_images)}, "
        f"Validation={len(validation_images)}, "
        f"Test={len(test_images)}"
    )

print("\nDataset split completed.")
```

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2

IMG_SIZE = (224, 224)

base_model = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

print("MobileNetV2 loaded successfully.")
print("Number of layers:", len(base_model.layers))
```

```python
import tensorflow as tf

base_path = "/content/drive/MyDrive/Crop_Classification/datasets/final_dataset_split"

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    f"{base_path}/train",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=42
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    f"{base_path}/validation",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    f"{base_path}/test",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

print("\nClasses:", train_ds.class_names)
```

```python
from tensorflow.keras import layers
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
])

# Apply augmentation only to training data
train_ds_augmented = train_ds.map(
    lambda x, y: (data_augmentation(x, training=True), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

# Apply MobileNetV2 preprocessing
train_ds_processed = train_ds_augmented.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

val_ds_processed = val_ds.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

test_ds_processed = test_ds.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

# Improve input pipeline performance
train_ds_processed = train_ds_processed.prefetch(tf.data.AUTOTUNE)
val_ds_processed = val_ds_processed.prefetch(tf.data.AUTOTUNE)
test_ds_processed = test_ds_processed.prefetch(tf.data.AUTOTUNE)

print("Preprocessing and augmentation ready.")
```

```python
from tensorflow.keras import layers, models

# Freeze the pretrained MobileNetV2 layers
base_model.trainable = False

# Build classification model
model = models.Sequential([
    base_model,

    layers.GlobalAveragePooling2D(),

    layers.Dropout(0.2),

    layers.Dense(128, activation="relu"),

    layers.Dropout(0.2),

    layers.Dense(5, activation="softmax")
])

model.summary()
```

```python
from tensorflow.keras.optimizers import Adam

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully.")
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

model_path = "/content/drive/MyDrive/Crop_Classification/best_mobilenetv2.keras"

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

checkpoint = ModelCheckpoint(
    model_path,
    monitor="val_loss",
    save_best_only=True,
    mode="min"
)

print("Callbacks ready.")
```

```python
EPOCHS = 20

history = model.fit(
    train_ds_processed,
    validation_data=val_ds_processed,
    epochs=EPOCHS,
    callbacks=[early_stopping, checkpoint]
)
```

```python
from tensorflow.keras.models import load_model

best_model = load_model(
    "/content/drive/MyDrive/Crop_Classification/best_mobilenetv2.keras"
)

print("Best model loaded successfully.")
```

```python
test_loss, test_accuracy = best_model.evaluate(
    test_ds_processed,
    verbose=1
)

print(f"\nTest Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")
```

```python
import numpy as np
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# Get true labels
y_true = np.concatenate([
    y.numpy() for x, y in test_ds_processed
])

# Get predictions
y_prob = best_model.predict(test_ds_processed)
y_pred = np.argmax(y_prob, axis=1)

class_names = test_ds.class_names

# Classification report
print("Classification Report:\n")
print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        digits=4
    )
)

# Confusion matrix
cm = confusion_matrix(y_true, y_pred)

print("\nConfusion Matrix:\n")
print(cm)

# Plot confusion matrix
plt.figure(figsize=(8, 6))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

disp.plot(
    cmap="Blues",
    values_format="d"
)

plt.title("Crop Classification - Confusion Matrix")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import tensorflow as tf

class_names = ["Maize", "Paddy", "Sugarcane", "Sunflower", "Wheat"]

def predict_crop(image_path):

    # Load and resize image
    image = tf.keras.utils.load_img(
        image_path,
        target_size=(224, 224)
    )

    # Convert image to array
    image_array = tf.keras.utils.img_to_array(image)

    # Add batch dimension
    image_array = np.expand_dims(image_array, axis=0)

    # MobileNetV2 preprocessing
    image_array = tf.keras.applications.mobilenet_v2.preprocess_input(
        image_array
    )

    # Predict
    predictions = best_model.predict(
        image_array,
        verbose=0
    )

    predicted_index = np.argmax(predictions[0])
    predicted_class = class_names[predicted_index]
    confidence = predictions[0][predicted_index] * 100

    print(f"Predicted Crop: {predicted_class}")
    print(f"Confidence: {confidence:.2f}%")

    return predicted_class, confidence

print("Prediction function ready.")
```

```python
prediction_script = r'''
import sys
import os
import numpy as np
import tensorflow as tf

# Get the folder where predict.py is located
SCRIPT_DIR = os.path.dirname(os.path.abspath(__file__))

MODEL_PATH = os.path.join(
    SCRIPT_DIR,
    "..",
    "best_mobilenetv2.keras"
)

CLASS_NAMES = [
    "Maize",
    "Paddy",
    "Sugarcane",
    "Sunflower",
    "Wheat"
]

IMG_SIZE = (224, 224)


def predict_crop(image_path):

    model = tf.keras.models.load_model(MODEL_PATH)

    image = tf.keras.utils.load_img(
        image_path,
        target_size=IMG_SIZE
    )

    image_array = tf.keras.utils.img_to_array(image)
    image_array = np.expand_dims(image_array, axis=0)

    image_array = tf.keras.applications.mobilenet_v2.preprocess_input(
        image_array
    )

    predictions = model.predict(
        image_array,
        verbose=0
    )

    predicted_index = np.argmax(predictions[0])
    predicted_class = CLASS_NAMES[predicted_index]
    confidence = predictions[0][predicted_index] * 100

    print(f"Predicted Crop: {predicted_class}")
    print(f"Confidence: {confidence:.2f}%")


if __name__ == "__main__":

    if len(sys.argv) != 2:
        print("Usage: python predict.py <image_path>")
        sys.exit(1)

    image_path = sys.argv[1]

    if not os.path.isfile(image_path):
        print(f"Error: Image not found: {image_path}")
        sys.exit(1)

    predict_crop(image_path)
'''

script_path = "/content/drive/MyDrive/Crop_Classification/prediction/predict.py"

with open(script_path, "w") as f:
    f.write(prediction_script)

print("Prediction script updated successfully.")
```
